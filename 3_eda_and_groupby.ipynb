# Exploratory Data Analysis and Groupby Insights

This notebook explores the cleaned dataset through summary statistics and groupby-based aggregations to identify patterns in abandonment behavior.


```python
import pandas as pd

df = pd.read_csv('ecommerce_cart_data_cleaned.csv')

overall_rate = (df['checkout_status'] == 'Abandoned').mean() * 100
print(f'Overall abandonment rate: {overall_rate:.2f}%')

device_abandonment = (
    df.groupby('device_type')['checkout_status']
    .apply(lambda status: (status == 'Abandoned').mean() * 100)
    .round(2)
    .reset_index(name='abandonment_rate_pct')
    .sort_values(by='abandonment_rate_pct', ascending=False)
)

hourly_abandonment = (
    df.groupby('hour_of_day')['checkout_status']
    .apply(lambda status: (status == 'Abandoned').mean() * 100)
    .round(2)
    .reset_index(name='abandonment_rate_pct')
)

print('\nAbandonment rate by device type:')
print(device_abandonment.to_string(index=False))

print('\nAbandonment rate by hour of day:')
print(hourly_abandonment.to_string(index=False))
```
