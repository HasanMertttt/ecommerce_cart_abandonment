# Data Cleaning

This notebook focuses on transforming the raw dataset into a reliable analytical dataset by addressing missing values, standardizing formats, and creating additional fields for downstream analysis.


```python
import pandas as pd

raw_df = pd.read_csv('ecommerce_cart_data_raw.csv')

median_cart_val = raw_df['cart_value'].median()
raw_df['cart_value'] = raw_df['cart_value'].fillna(median_cart_val)
raw_df['device_type'] = raw_df['device_type'].fillna('Unknown')
raw_df['session_start_time'] = pd.to_datetime(raw_df['session_start_time'])
raw_df['hour_of_day'] = raw_df['session_start_time'].dt.hour
raw_df['day_of_week'] = raw_df['session_start_time'].dt.day_name()

raw_df.to_csv('ecommerce_cart_data_cleaned.csv', index=False)
print('Cleaning complete. Cleaned dataset saved to ecommerce_cart_data_cleaned.csv')
raw_df.head()
```
