# Advanced Visualizations

This notebook translates the analysis into clear, presentation-ready visuals that highlight how abandonment rates vary across time and device type.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap

plt.style.use('dark_background')
bg_color = '#0e1117'
plt.rcParams.update({
    'figure.facecolor': bg_color,
    'axes.facecolor': bg_color,
    'savefig.facecolor': bg_color,
    'grid.color': '#21262d',
    'font.family': 'sans-serif',
    'font.size': 10,
    'text.color': '#e6edf3',
    'axes.labelcolor': '#8b949e',
    'xtick.color': '#8b949e',
    'ytick.color': '#8b949e',
})

colors = ["#0e1117", "#2c1b4d", "#6b21a8", "#a855f7", "#e879f9"]
neon_purple_cmap = LinearSegmentedColormap.from_list("NeonPurpleBlack", colors)

df = pd.read_csv('ecommerce_cart_data_cleaned.csv')
df['is_abandoned'] = (df['checkout_status'] == 'Abandoned').astype(int)

heatmap_data = (
    df.pivot_table(index='hour_of_day', columns='day_of_week', values='is_abandoned', aggfunc='mean') * 100
)
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
heatmap_data = heatmap_data.reindex(columns=days_order)

fig, ax = plt.subplots(figsize=(10, 8), dpi=300)
sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='.1f',
    cmap=neon_purple_cmap,
    linewidths=0.5,
    linecolor='#161b22',
    cbar_kws={'label': 'Abandonment Rate (%)'},
    ax=ax
)
ax.set_title('Cart Abandonment Rate Heatmap (Hour vs Day)', fontsize=14, fontweight='bold', pad=15, color='#f0f6fc')
ax.set_xlabel('Day of Week', fontsize=11, labelpad=10)
ax.set_ylabel('Hour of Day (00:00 - 23:00)', fontsize=11, labelpad=10)
plt.tight_layout()
plt.savefig('cart_abandonment_heatmap.png', dpi=300, bbox_inches='tight')
plt.close()

device_df = (
    df.groupby('device_type')['is_abandoned']
    .mean()
    .mul(100)
    .reset_index(name='abandonment_rate')
    .sort_values(by='abandonment_rate', ascending=False)
)

fig, ax = plt.subplots(figsize=(8, 5), dpi=300)
neon_cyan = '#00f5d4'
bars = ax.bar(device_df['device_type'], device_df['abandonment_rate'], color=neon_cyan, edgecolor='#00e5ff', linewidth=1.2, width=0.5)
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}%', xy=(bar.get_x() + bar.get_width() / 2, height), xytext=(0, 5), textcoords='offset points', ha='center', va='bottom', fontsize=11, fontweight='bold', color='#ffffff')

ax.set_title('Cart Abandonment Rate by Device Type', fontsize=14, fontweight='bold', pad=15, color='#f0f6fc')
ax.set_xlabel('Device Type', fontsize=11, labelpad=10)
ax.set_ylabel('Abandonment Rate (%)', fontsize=11, labelpad=10)
ax.set_ylim(0, 80)
plt.tight_layout()
plt.savefig('cart_abandonment_device_barchart.png', dpi=300, bbox_inches='tight')
plt.close()

print('Visualizations generated successfully.')
```
